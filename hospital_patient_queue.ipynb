```python
class PatientNode:
    def __init__(self, name):
        self.name = name
        self.next = None


class PatientQueue:
    def __init__(self):
        self.front = None
        self.rear = None

    # Add patient
    def enqueue(self, name):
        new_patient = PatientNode(name)

        if self.rear is None:
            self.front = self.rear = new_patient
        else:
            self.rear.next = new_patient
            self.rear = new_patient

        print("Patient added:", name)

    # Remove patient
    def dequeue(self):
        if self.front is None:
            print("Queue is empty")
        else:
            print("Patient treated:", self.front.name)
            self.front = self.front.next

            if self.front is None:
                self.rear = None

    # View front patient
    def peek(self):
        if self.front is None:
            print("Queue is empty")
        else:
            print("Front patient:", self.front.name)

    # Display all patients
    def display(self):
        if self.front is None:
            print("Queue is empty")
        else:
            temp = self.front

            print("Patients in queue:")

            while temp is not None:
                print(temp.name)
                temp = temp.next

    # Count patients
    def count(self):
        count = 0
        temp = self.front

        while temp is not None:
            count += 1
            temp = temp.next

        print("Total patients:", count)

queue = PatientQueue()

n = int(input("Enter number of patients: "))

for i in range(n):
    name = input("Enter patient name: ")
    queue.enqueue(name)

print()

queue.display()

queue.count()

queue.peek()

print()

queue.dequeue()

print()

queue.display()

queue.count()
queue.peek()
```
